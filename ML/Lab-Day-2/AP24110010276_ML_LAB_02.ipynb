Column Addition

```python

import pandas as pd
info = {'one' : pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e']),
'two' : pd.Series([1, 2, 3, 4, 5, 6], index=['a', 'b', 'c', 'd', 'e', 'f'])}
df = pd.DataFrame(info)
print(df)
# Add a new column to an existing DataFrame object
print ("Add new column by passing series")
df['three']=pd.Series([20,40,60],index=['a','b','c'])
print (df)
print ("Add new column using existing DataFrame columns")
df['four']=df['one']+df['three']
print (df)
```

Deletion:

```python

import pandas as pd

info = {'one' : pd.Series([1, 2], index= ['a', 'b']),
'two' : pd.Series([1, 2, 3], index=['a', 'b', 'c'])}

df = pd.DataFrame(info)
print ("The DataFrame:")
print (df)

# using del function
print ("Delete the first column:")
del df['one']
print (df)
# using pop function
print ("Delete the another column:")
df.pop('two')
print (df)
```

Selection by Label:

```python

# We can select any row by passing the row label to a loc function.
import pandas as pd
info = {'one' : pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e']),
'two' : pd.Series([1, 2, 3, 4, 5, 6], index=['a', 'b', 'c', 'd', 'e', 'f'])}

df = pd.DataFrame(info)
print("Data frame:")
print(df)
print("b index elements are:")
print (df.loc['b'])
print("e index elements are:")
print(df.loc['e'])
```

Slice Rows
It is another method to select multiple rows using ':' operator.

```python

import pandas as pd
info = {'one' : pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e']),
'two' : pd.Series([1, 2, 3, 4, 5, 6], index=['a', 'b', 'c', 'd', 'e', 'f'])}
df = pd.DataFrame(info)
print(df)
print (df[2:5])
```

Addition of rows:

We can easily add new rows to the DataFrame using append function. It adds the
new rows at the end.

```python

import pandas as pd

# Assuming d and d2 are defined DataFrames
d = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})
d2 = pd.DataFrame({'A': [7, 8, 9], 'B': [10, 11, 12]})

# Replace append with concat
d = pd.concat([d, d2])

print(d)
```

Deletion of rows:
We can delete or drop any rows from a DataFrame using the index label. If in case,
the label is duplicate then multiple rows will be deleted.

```python


a_info = pd.DataFrame([[4, 5], [6, 7]], columns = ['x','y'])
b_info = pd.DataFrame([[8, 9], [10, 11]], columns = ['x','y'])
a_info = pd.concat([a_info,b_info])
print( 'Drop the row with label 0')
# Drop rows with label 0
a_info = a_info.drop(0)

print(a_info)
```

DataFrame.drop_duplicates()

```python

import pandas as pd
emp = {"Name": ["Siva", "Rama", "Krishna", "Siva"],
"Age": [21, 32, 29, 21]}
info = pd.DataFrame(emp)
print(info)
info = info.drop_duplicates()
print("After removing the duplicates the dataframe is :")
print(info)
```

Null values identification in given dataset:

```python

import pandas as pd
import numpy as np
info = pd.DataFrame({"Person":["Siva", "Pavan", "Gopal", "Raghu"],
"Age": [27., 29, np.nan, 32]})
print(info)
print(info.count())
print("null values")
print(info.isnull())
print(info.isnull().sum())
```

Sorting order

```python

import pandas as pd
import numpy as np
info = pd.DataFrame({"Person":["Siva", "Pavan", "Gopal", "Raghu"],
"Age": [27., 29, np.nan, 32]})
print(info)
print("Sorting Order")
print(info.sort_index())
print("Descending Order")
print(info.sort_index(ascending=False))
```
